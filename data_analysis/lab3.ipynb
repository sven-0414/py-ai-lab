```python
loans = loans_raw.copy()

loans['genre'] = (loans['genre']
    .str.strip()
    .str.lower()
    .replace({"children's": 'children', 'sci fi': 'science fiction', 'sci-fi': 'science fiction'})
    .str.title())

loans['returned'] = (loans['returned']
    .str.strip()
    .str.lower()
    .map({'y': True, 'yes': True, 'n': False, 'no': False}))

loans['late_fee'] = (loans['late_fee']
    .str.replace('kr', '')
    .str.replace(' ', '')
    .str.replace(',', '.')
    .astype(float))

loans['loan_date'] = pd.to_datetime(loans['loan_date'], errors='coerce')

loans = loans.drop(columns='notes')

loans = loans.drop_duplicates().reset_index(drop=True)

loans.to_csv('loans_clean.csv', index=False)

print(loans.shape)
```
