```python
import numpy as np
import pandas as pd

listings = pd.read_csv('data/listings.csv')
listings.shape
```

```python
listings.info()
```

```python
listings['property_type'].unique()
```

```python
listings['property_type'].value_counts(normalize=True).head()
```

```python
listings['neighbourhood_cleansed'].value_counts(ascending=True).head()
```

```python
listings.groupby('room_type')['beds'].mean().round(2)
```

```python
listings.groupby('room_type')['beds'].agg(['count', 'mean']).round(2)
```

```python
listings.groupby('neighbourhood_cleansed')['number_of_reviews'].agg(['count', 'mean', 'median']).round(2)
```

```python
listings.groupby('room_type').agg(
    n_listings=('id', 'count'),
    avg_beds=('beds', 'mean'),
    max_reviews=('number_of_reviews', 'max'),
).round(2)
```

```python
(pd.crosstab(listings['neighbourhood_cleansed'], listings['host_is_superhost'], normalize='index') * 100).round(1).sort_values(by='t', ascending=False)
```

```python
listings['host_is_superhost'].unique()
```

```python
listings['stay'] = pd.cut(listings['minimum_nights'], bins=[0,2,6,29,500])
listings['stay'].value_counts().sort_index()
```
